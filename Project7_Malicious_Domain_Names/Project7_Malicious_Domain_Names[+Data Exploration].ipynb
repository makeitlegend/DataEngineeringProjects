##Data Exploration

```python
df_malicious_domains = pd.read_csv("/content/Malicious Domain Name Dataset_10000.csv")
```

```python
df_malicious_domains.info()
```

```python
df_malicious_domains.describe(include='all')
```

```python
columns_malicious_domains = [
    'domain_name',
    'asn',
    'ip',
    'owner',
    'country_code',
    'malware',
    'domain _length',
    'n_letters',
    'n_number',
    'n_hypans',
    'n_dots',
    'entropy'
]

for col in columns_malicious_domains:
    missing = df_malicious_domains[col].isna().sum()
    duplicated = df_malicious_domains[col].duplicated().sum()
    total_non_null = df_malicious_domains[col].count()

    print(f"--- {col} ---")
    print(f"Missing values:        {missing}")
    print(f"Duplicated values:     {duplicated}")
    print(f"Total non-null values: {total_non_null}\n")
```
