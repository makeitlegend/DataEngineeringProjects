##Data Exploration

```python
df_yt_tech_channels = pd.read_csv("/content/youtube_tech_channels.csv")

df_yt_tech_vids = pd.read_csv("/content/youtube_tech_videos.csv")
```

```python
df_yt_tech_channels.info()

df_yt_tech_vids.info()
```

```python
'''
------------------------------------
Column Descriptions (source: Kaggle)
------------------------------------
Channels:
channel_id: Unique YouTube channel identifier
channel_name: Display name of the channel
description: Channel description (first 500 chars)
subscribers: Number of subscribers
total_views: Lifetime view count
total_videos: Total videos published
created_date: Channel creation date
country: Channel's country (if available)
scraped_at: Data collection timestamp

Videos:
video_id: Unique YouTube video identifier
title: Video title
published_at: Upload timestamp
views: View count
likes: Like count
comments: Comment count
duration: Video duration (ISO 8601 format)
thumbnail: High-quality thumbnail URL
video_url: Direct link to video
channel_id: Parent channel ID
channel_name: Parent channel name
scraped_at: Data collection timestamp
'''
```

```python
df_yt_tech_channels.describe(include='all')
```

```python
df_yt_tech_vids.describe(include='all')
```

```python
columns_yt_tech_channels = [
    'channel_id', 'channel_name', 'description', 'subscribers',
    'total_views', 'total_videos', 'created_date', 'country',
    'scraped_at'
]

for col in columns_yt_tech_channels:
    missing = df_yt_tech_channels[col].isna().sum()
    duplicated = df_yt_tech_channels[col].duplicated().sum()
    total_non_null = df_yt_tech_channels[col].count()

    print(f"--- {col} ---")
    print(f"Missing values:        {missing}")
    print(f"Duplicated values:     {duplicated}")
    print(f"Total non-null values: {total_non_null}\n")
```

```python
columns_yt_tech_vids = [
    'video_id', 'title', 'published_at', 'views', 'likes', 'comments',
    'duration', 'thumbnail', 'video_url', 'channel_id',
    'channel_name', 'scraped_at'
]

for col in columns_yt_tech_vids:
    missing = df_yt_tech_vids[col].isna().sum()
    duplicated = df_yt_tech_vids[col].duplicated().sum()
    total_non_null = df_yt_tech_vids[col].count()

    print(f"--- {col} ---")
    print(f"Missing values:        {missing}")
    print(f"Duplicated values:     {duplicated}")
    print(f"Total non-null values: {total_non_null}\n")
```
