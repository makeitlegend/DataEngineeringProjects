##Data Exploration

```python
df_cyber_gaming = pd.read_csv("/content/Cyber vigilance dataset.csv")
```

```python
df_cyber_gaming.info()
```

```python
df_cyber_gaming.describe(include='all')
```

```python
columns_cyber_gaming = [
    'Nickname',
    'DefenderScore',
    'AttackerScore',
    'Time(sec)',
    'Winner',
    'Level'
]

for col in columns_cyber_gaming:
    missing = df_cyber_gaming[col].isna().sum()
    duplicated = df_cyber_gaming[col].duplicated().sum()
    total_non_null = df_cyber_gaming[col].count()

    print(f"--- {col} ---")
    print(f"Missing values:        {missing}")
    print(f"Duplicated values:     {duplicated}")
    print(f"Total non-null values: {total_non_null}\n")
```
